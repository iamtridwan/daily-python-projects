```python
orders_df = pd.read_csv('orders_cleaned.csv')
```

```python
orders_df.head()
orders_df['total_price'] = orders_df['quantity'] * orders_df["unit_price"]
```

```python
orders_df.info()
```

```python
completed = orders_df[orders_df["status"] == 'Completed']

category_revenue = completed.groupby("category")["total_price"].sum().sort_values()

category_revenue.max()
```
